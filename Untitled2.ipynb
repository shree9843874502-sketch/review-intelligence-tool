```python
from google_play_scraper import reviews, Sort
import pandas as pd

APP_ID = "in.swiggy.android"   # unga app-oda ID inga maathunga

result, _ = reviews(APP_ID, lang="en", country="in",
                    sort=Sort.NEWEST, count=500)
df = pd.DataFrame(result)[["at", "score", "content"]]
df.columns = ["date", "rating", "text"]
df.to_csv("reviews.csv", index=False)
df.head()
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
sia = SentimentIntensityAnalyzer()
df["sentiment"] = df["text"].apply(lambda t: sia.polarity_scores(str(t))["compound"])
df["label"] = df["sentiment"].apply(
    lambda s: "positive" if s > 0.05 else "negative" if s < -0.05 else "neutral")
df["label"].value_counts()
```

```python
import re
from collections import Counter

neg = df[df["label"] == "negative"]["text"].astype(str).str.lower()

stop = set("the a an and or to of in is it i my me for on with this that was are but not so very have has had you your app swiggy they be as at from just get got can will would if there when what which do does did dont don't no".split())

words = []
for t in neg:
    words += [w for w in re.findall(r"[a-z']+", t) if w not in stop and len(w) > 2]

Counter(words).most_common(20)
```

```python
themes = {
    "delivery delay": r"\b(late|delay|delayed|slow|waiting)\b",
    "refund / money": r"\b(refund|money|charged|deduct|payment)\b",
    "support": r"\b(support|customer care|service|response)\b",
    "app issues": r"\b(crash|bug|hang|glitch|error|login)\b",
    "price": r"\b(expensive|costly|price|charges|fee)\b",
    "order problems": r"\b(cancel|wrong|missing|cold|stale)\b",
}

neg_all = df[df["label"] == "negative"]["text"].astype(str).str.lower()
counts = {k: neg_all.str.contains(p, regex=True).sum() for k, p in themes.items()}
pd.Series(counts).sort_values(ascending=False)
```

```python
neg_df = df[df["label"] == "negative"]
sup = neg_df[neg_df["text"].astype(str).str.lower().str.contains(r"\b(?:support|customer care|service|response)\b", regex=True)]
for t in sup["text"].head(10):
    print("-", t[:200], "\n")
```

```python
import matplotlib.pyplot as plt
s = pd.Series(counts).sort_values()
s.plot(kind="barh", title="Top complaints in negative reviews")
plt.xlabel("Number of reviews")
plt.tight_layout()
plt.savefig("complaints.png", dpi=150)
plt.show()
```
